## Data Visualization

### Distribution of Annual Salary

```python
plt.figure(figsize=(8,5))
sns.histplot(df["annual_salary_usd"], bins=38)
plt.title("Distribution of Annual Salary")
plt.xlabel("Salary (USD)")
plt.ylabel("Frequency")
plt.show()
```

Most AI job salaries are concentrated between 120,000 USD and 250,000 USD, indicating a mid-to-high salary range in the industry.

### Experience vs Salary Relationship

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="years_of_experience", y="annual_salary_usd", data=df)
plt.title("Experience vs Salary")
plt.xlabel("Years of experience")
plt.ylabel("Salary (USD)")
plt.show()
```

The scatter plot shows a general upward trend, indicating that salary tends to increase with years of experience.

### Job Category Distribution

```python
plt.figure(figsize=(10,5))
sns.countplot(x="job_category", data=df)
plt.title("Job Category Distribition")
plt.xticks(rotation=45)
plt.show
```

AI Engineering roles dominate the dataset, showing higher demand compared to other job categories.

### Remote Work Distribution

```python
sns.countplot(x="remote_work", data=df)
plt.title("Remote Work Distribution")
plt.show()
```

Hybrid work is the most common work model, followed by fully remote and on-site roles.

### Average Salary by Job Category

```python
plt.figure(figsize=(10,5))
sns.barplot(x="job_category", y="annual_salary_usd", data=df)
plt.title("Average Salary (USD)")
plt.xlabel("Job Category")
plt.ylabel("Average Salary (USD)")
plt.xticks(rotation=45)
plt.show()
```

Certain roles like Architecture and Infrastructure offer higher average salaries compared to other job categories.

### Salary Distribution by Remote Work Type

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="remote_work", y="annual_salary_usd", data=df)
plt.title("Salary Distribution by Remote Work Type")
plt.xlabel("Remote Work Type")
plt.ylabel("Salary (USD)")
plt.show()
```

Salary distribution across remote, hybrid, and on-site roles is relatively similar, with slight variations in higher salary ranges.

## Key Insights

1. Most AI job salaries fall between 120,000 USD and 250,000 USD.
2. Salary generally increases with years of experience.
