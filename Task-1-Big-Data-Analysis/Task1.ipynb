```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("JobDataAnalysis").getOrCreate()

df = spark.read.csv("Dataset.csv", header=True, inferSchema=True)
df.show(5)
```

```python
df.printSchema()
df.describe().show()
```

```python
df.groupBy("job_category").count().show()
```

```python
df.groupBy("job_title").avg("annual_salary_usd").show()
```

```python
df.groupBy("experience_level").avg("annual_salary_usd").show()
```

```python
df.groupBy("country").avg("annual_salary_usd").show()
```

```python
df.groupBy("remote_work").count().show()
```

```python
df.groupBy("company_size").avg("annual_salary_usd").show()
```

```python
df.filter(df.annual_salary_usd > 200000).show()
```

```python
from pyspark.sql.functions import when

df = df.withColumn(
    "salary_level",
    when(df.annual_salary_usd > 200000, "High")
    .when(df.annual_salary_usd > 100000, "Medium")
    .otherwise("Low")
)

df.select("job_title", "salary_level").show()
```
